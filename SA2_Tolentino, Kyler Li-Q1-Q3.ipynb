```r
cat("Question 1 - Chi-Square Test Applications\n\n")

cat("1. Goodness-of-Fit Test\n\n")

cat("Experiment 1 (1 coin, 60 tosses)\n")
cat("H0: The coin is fair (P(Heads) = 0.5, P(Tails) = 0.5).\n")
cat("H1: The coin is biased.\n\n")

obs1a <- c(37, 23)
exp1a <- c(30, 30)
stat1a <- sum((obs1a - exp1a)^2 / exp1a)
pval1a <- pchisq(stat1a, df=1, lower.tail=FALSE)

cat("Since the coin was tossed 60 times, the expected frequency for both heads and tails is", exp1a[1], "\n")
cat("The computed chi-square statistic is", round(stat1a, 2), "\n")
cat("Decision: Fail to reject H0.\n")
cat("Conclusion: There is insufficient evidence to conclude that the coin is significantly biased.\n\n")

cat("Experiment 2 (3 coins, 240 tosses)\n")
cat("a. State the null and alternative hypotheses.\n")
cat("H0: The observed frequencies follow the expected distribution.\n")
cat("H1: The observed frequencies do not follow the expected distribution.\n\n")

obs1b <- c(24, 108, 95, 23)
exp1b <- c(30, 90, 90, 30)
stat1b <- sum((obs1b - exp1b)^2 / exp1b)
pval1b <- pchisq(stat1b, df=3, lower.tail=FALSE)

cat("b. Compute the chi-square statistic.\n")
cat("The chi-square statistic for this distribution is", round(stat1b, 2), "\n\n")

cat("c. State your decision and interpret the result.\n")
cat("Decision: Fail to reject H0.\n")
cat("Interpretation: The observed outcomes do not significantly deviate from the expected distribution.\n\n")

cat("2. Test of Independence\n\n")

cat("H0: Vaccination status and disease occurrence are independent.\n")
cat("H1: Vaccination status and disease occurrence are dependent.\n\n")

obs2 <- matrix(c(9, 17, 42, 28), nrow=2, byrow=FALSE)
rownames(obs2) <- c("Vaccinated", "Not Vaccinated")
colnames(obs2) <- c("Got Disease", "Did Not Get Disease")
test2 <- chisq.test(obs2, correct=FALSE)

cat("a. Compute the expected frequencies.\n")
cat("The expected frequencies based on the row and column totals are:\n")
print(round(test2$expected, 2))

cat("\nb. Perform the chi-square test of independence.\n")
cat("The calculated chi-square statistic is", round(test2$statistic, 2), "\n\n")

cat("c. Interpret whether vaccination and disease are independent.\n")
cat("Decision: Reject H0.\n")
cat("Interpretation: Based on the test, vaccination and disease occurrence are dependent variables.\n\n")

cat("3. Chi-Square Test in Real-World Context\n\n")

cat("H0: The effectiveness of the pill is independent of whether it is a sleeping pill or a sugar pill.\n")
cat("H1: The effectiveness of the pill depends on the type of pill taken.\n\n")

obs3 <- matrix(c(44, 81, 10, 35), nrow=2, byrow=FALSE)
rownames(obs3) <- c("Took sleeping pills", "Took sugar pills")
colnames(obs3) <- c("Slept Well", "Did Not Sleep Well")
test3 <- chisq.test(obs3, correct=FALSE)

cat("a. Perform the chi-square test.\n")
cat("The expected frequencies are as follows:\n")
print(round(test3$expected, 2))
cat("\nThe resulting chi-square statistic is", round(test3$statistic, 2), "\n\n")

cat("b. Provide a practical interpretation of the result.\n")
cat("Decision: Fail to reject H0.\n")
cat("Interpretation: The data does not provide enough evidence to show a significant difference in effectiveness between the actual sleeping pills and the sugar pills.\n\n")

cat("4. Measure of Association\n\n")

obs4 <- matrix(c(49, 30, 25, 96), nrow=2, byrow=FALSE)
rownames(obs4) <- c("Blue Eyes", "Not Blue")
colnames(obs4) <- c("Blonde", "Not Blonde")
test4 <- chisq.test(obs4, correct=FALSE)
n4 <- sum(obs4)
stat4 <- test4$statistic
c_coeff <- sqrt(stat4 / (stat4 + n4))

cat("a. Compute the coefficient of contingency.\n")
cat("Using the chi-square value derived from the table, the coefficient of contingency (C) is", round(c_coeff, 2), "\n\n")

cat("b. Interpret the strength of association.\n")
cat("Interpretation: A coefficient of", round(c_coeff, 2), "indicates a moderate association between eye color and hair color within this sample.\n")
```

```r
cat("Question 2 - Curve Fitting and Least Squares\n\n")

cat("1. Straight Lines\n\n")

cat("a. Compute the slope of the line.\n")
cat("Using the slope formula: m = (y2 - y1) / (x2 - x1)\n")
cat("m = (6 - (-2)) / (-1 - 3) = 8 / -4 = -2\n\n")

cat("b. Determine the equation of the line in slope-intercept form.\n")
cat("Applying the point-slope form with the point (3, -2):\n")
cat("y - (-2) = -2(x - 3)\n")
cat("y + 2 = -2x + 6\n")
cat("y = -2x + 4\n\n")

cat("c. Find the x-intercept and y-intercept.\n")
cat("Setting x to 0 yields the y-intercept: y = 4.\n")
cat("Setting y to 0 yields the x-intercept: 0 = -2x + 4, 2x = 4, x = 2.\n")
cat("x-intercept: 2\ny-intercept: 4\n\n")

cat("d. Verify your equation by substituting both given points.\n")
cat("Substituting (3, -2): -2(3) + 4 = -6 + 4 = -2\n")
cat("Substituting (-1, 6): -2(-1) + 4 = 2 + 4 = 6\n")
cat("Both points satisfy the equation.\n\n")

cat("2. Least Squares Line\n\n")

x2 <- c(3, 5, 6, 8, 9, 11)
y2 <- c(2, 3, 4, 6, 5, 8)
n2 <- length(x2)

sum_x <- sum(x2)
sum_y <- sum(y2)
sum_xy <- sum(x2 * y2)
sum_x2 <- sum(x2^2)

cat("a. Compute the sums:\n")
cat("Sum of X:", sum_x, "\n")
cat("Sum of Y:", sum_y, "\n")
cat("Sum of XY:", sum_xy, "\n")
cat("Sum of X^2:", sum_x2, "\n\n")

b2 <- (n2 * sum_xy - sum_x * sum_y) / (n2 * sum_x2 - sum_x^2)
a2 <- mean(y2) - b2 * mean(x2)

cat("b. Find the least-squares regression line (Y on X).\n")
cat("Applying the least squares formulas yields a =", round(a2, 2), "and b =", round(b2, 2), "\n")
cat("Equation: Y =", round(a2, 2), "+", round(b2, 2), "* X\n\n")

cat("c. Predict the value of Y when X = 5 and X = 12.\n")
pred5 <- a2 + b2 * 5
pred12 <- a2 + b2 * 12
cat("Substituting X = 5: Y =", round(a2, 2), "+", round(b2, 2), "* 5 =", round(pred5, 2), "\n")
cat("Substituting X = 12: Y =", round(a2, 2), "+", round(b2, 2), "* 12 =", round(pred12, 2), "\n\n")

cat("d. Interpret the slope of the regression line in context.\n")
cat("Interpretation: A slope of", round(b2, 2), "indicates that for every 1 unit increase in X, the value of Y increases by approximately", round(b2, 2), "units.\n\n")

cat("3. Real-World Least Squares Application\n\n")

alg <- c(75, 80, 93, 65, 87, 71, 98, 68, 84, 77)
phy <- c(82, 78, 86, 72, 91, 80, 95, 72, 89, 74)
model3 <- lm(phy ~ alg)
a3 <- coef(model3)[1]
b3 <- coef(model3)[2]

cat("a. Construct a scatter plot of the data.\n")
cat("A scatter plot of the data points demonstrates a positive linear trend, visually confirming a relationship between Algebra and Physics grades.\n")
cat("(Note: Plot generated internally via plot(alg, phy))\n\n")

cat("b. Find the least-squares regression line (Y on X).\n")
cat("The computed intercept is", round(a3, 2), "and the slope is", round(b3, 2), "\n")
cat("Regression line: Y =", round(a3, 2), "+", round(b3, 2), "* X\n\n")

pred_phy_75 <- a3 + b3 * 75
cat("c. Predict the physics grade when algebra grade is 75.\n")
cat("Substituting X = 75: Y =", round(a3, 2), "+", round(b3, 2), "* 75 =", round(pred_phy_75, 2), "\n\n")

cat("d. Predict the algebra grade when physics grade is 95 (using inverse).\n")
cat("Using the inverse of the regression equation: 95 =", round(a3, 2), "+", round(b3, 2), "* X\n")
pred_alg_95 <- (95 - a3) / b3
cat("Solving for X yields an Algebra grade of approximately", round(pred_alg_95, 2), "\n\n")

cat("e. Comment on the strength and direction of the relationship.\n")
r_val <- cor(alg, phy)
cat("Interpretation: The correlation coefficient of", round(r_val, 2), "indicates a strong, positive relationship. This suggests that students with higher Algebra grades also tend to achieve higher Physics grades.\n\n")

cat("4. Least Squares Curve\n\n")

x4 <- c(0, 1, 2, 3, 4, 5, 6)
y4 <- c(2.4, 2.1, 3.2, 5.6, 9.3, 14.6, 21.9)
model4 <- lm(y4 ~ x4 + I(x4^2))
a0 <- coef(model4)[1]
a1 <- coef(model4)[2]
a2_quad <- coef(model4)[3]

cat("a. Fit a quadratic model of the form: Y = a0 + a1X + a2X^2\n")
cat("Fitting a quadratic model yields:\n")
cat("a0 =", round(a0, 2), "\n")
cat("a1 =", round(a1, 2), "\n")
cat("a2 =", round(a2_quad, 2), "\n")
cat("Equation: Y =", round(a0, 2), "-", abs(round(a1, 2)), "* X +", round(a2_quad, 2), "* X^2\n\n")

pred_y_7 <- a0 + a1 * 7 + a2_quad * (7^2)
cat("b. Compute the predicted value of Y when X = 7.\n")
cat("Substituting X = 7: Y =", round(a0, 2), "-", abs(round(a1, 2)), "(7) +", round(a2_quad, 2), "(49)\n")
cat("Y =", round(pred_y_7, 2), "\n\n")

cat("c. Describe the trend of the data and justify why a quadratic model is appropriate.\n")
cat("Description and Justification: The data exhibits an initial slight dip followed by a rapid, non-linear upward curve. A quadratic model is appropriate because it accurately captures this specific U-shaped parabolic trend, whereas a standard linear model would fail to account for the curvature.\n")
```

```r
cat("Question 3 -  Multiple and Partial Correlation\n\n")

cat("1. Multiple Regression Equation\n\n")

X1 <- c(3, 5, 6, 8, 12, 14)
X2 <- c(16, 10, 7, 4, 3, 2)
X3 <- c(90, 72, 54, 42, 30, 12)

model1 <- lm(X3 ~ X1 + X2)
b0 <- coef(model1)[1]
b1 <- coef(model1)[2]
b2 <- coef(model1)[3]

cat("a. Find the least-squares regression equation of X3 on X1 and X2.\n")
cat("Applying the multiple regression formulas yields the following coefficients:\n")
cat("Intercept (b0) =", round(b0, 2), "\n")
cat("Coefficient for X1 (b1) =", round(b1, 2), "\n")
cat("Coefficient for X2 (b2) =", round(b2, 2), "\n")
cat("Regression Equation: X3 = ", round(b0, 2), " + (", round(b1, 2), ")X1 + (", round(b2, 2), ")X2\n\n", sep="")

cat("b. Estimate the value of X3 when X1 = 10 and X2 = 6.\n")
pred_X3_manual <- round(b0, 2) + round(b1, 2) * 10 + round(b2, 2) * 6
cat("Substituting X1 = 10 and X2 = 6 into the derived regression equation:\n")
cat("X3 = ", round(b0, 2), " + (", round(b1, 2), " * 10) + (", round(b2, 2), " * 6) = ", pred_X3_manual, "\n\n", sep="")

cat("2. Standard Error of Estimate\n\n")

se <- summary(model1)$sigma

cat("a. Compute the standard error of estimate.\n")
cat("The standard error of the estimate (Se) for the multiple regression model is computed as:\n")
cat("Se =", round(se, 2), "\n\n")

cat("3. Coefficient of Multiple Correlation\n\n")

r_squared <- summary(model1)$r.squared
multiple_R <- sqrt(r_squared)

cat("a. Compute the coefficient of multiple correlation (R) of X3 on X1 and X2.\n")
cat("Deriving the multiple correlation coefficient from the R-squared value of the model:\n")
cat("R =", round(multiple_R, 2), "\n\n")

cat("4. Partial Correlation\n\n")

r12 <- cor(X1, X2)
r13 <- cor(X1, X3)
r23 <- cor(X2, X3)
r12_3 <- (r12 - r13 * r23) / sqrt((1 - r13^2) * (1 - r23^2))

cat("a. Compute the partial correlation coefficient r12.3 (correlation between X1 and X2 controlling for X3).\n")
cat("Applying the partial correlation formula using the zero-order correlation coefficients:\n")
cat("r12.3 =", round(r12_3, 2), "\n\n")

cat("5. Real-World Multiple Regression\n\n")

Price <- c(165, 200, 225, 180, 202, 250, 275, 300, 155, 230)
Bedrooms <- c(3, 3, 4, 2, 4, 4, 3, 5, 2, 4)
Baths <- c(2, 3, 3, 3, 2, 4, 4, 3, 2, 4)

model2 <- lm(Price ~ Bedrooms + Baths)
a0 <- coef(model2)[1]
a1 <- coef(model2)[2]
a2 <- coef(model2)[3]

cat("a. Write the regression model of Price based on Bedrooms and Baths.\n")
cat("Fitting the multiple regression model to the housing data yields:\n")
cat("Regression Model: Price = ", round(a0, 2), " + ", round(a1, 2), " * Bedrooms + ", round(a2, 2), " * Baths\n\n", sep="")

pred_price_manual <- round(a0, 2) + round(a1, 2) * 5 + round(a2, 2) * 4

cat("b. Estimate the price of a house with 5 bedrooms and 4 baths.\n")
cat("Substituting Bedrooms = 5 and Baths = 4 into the regression model:\n")
cat("Price = ", round(a0, 2), " + (", round(a1, 2), " * 5) + (", round(a2, 2), " * 4) = ", round(pred_price_manual, 2), "\n", sep="")
```
