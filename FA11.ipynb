```r
data("PlantGrowth")

#Descriptive Statistics
groupSizes <- table(PlantGrowth$group)
groupMeans <- aggregate(weight ~ group, data = PlantGrowth, mean)
groupSDs <- aggregate(weight ~ group, data = PlantGrowth, sd)

descriptiveStats <- data.frame(
  Group = groupMeans$group,
  N = as.numeric(groupSizes),
  Mean = round(groupMeans$weight, 3),
  StdDeviation = round(groupSDs$weight, 3)
)
print("Descriptive Statistics")
write.table(descriptiveStats, sep = " | ", quote = FALSE, row.names = FALSE)
cat("\n")

#Normality Test
shapiroCtrl <- shapiro.test(PlantGrowth$weight[PlantGrowth$group == "ctrl"])
shapiroTrt1 <- shapiro.test(PlantGrowth$weight[PlantGrowth$group == "trt1"])
shapiroTrt2 <- shapiro.test(PlantGrowth$weight[PlantGrowth$group == "trt2"])

normalityResults <- data.frame(
  Group = c("ctrl", "trt1", "trt2"),
  ShapiroWilkW = round(c(shapiroCtrl$statistic, shapiroTrt1$statistic, shapiroTrt2$statistic), 3),
  Pvalue = round(c(shapiroCtrl$p.value, shapiroTrt1$p.value, shapiroTrt2$p.value), 3)
)
print("Normality Test")
write.table(normalityResults, sep = " | ", quote = FALSE, row.names = FALSE)
cat("\n")

#Equality of Variances
bartlettTest <- bartlett.test(weight ~ group, data = PlantGrowth)
varianceResults <- data.frame(
  Test = "Bartletts",
  Statistic = round(bartlettTest$statistic, 4),
  df = bartlettTest$parameter,
  p = round(bartlettTest$p.value, 4)
)
print("Equality of Variances")
write.table(varianceResults, sep = " | ", quote = FALSE, row.names = FALSE)
cat("\n")

#ANOVA Table
anovaModel <- aov(weight ~ group, data = PlantGrowth)
anovaTable <- summary(anovaModel)[[1]]

anovaPrint <- data.frame(
  Cases = c("Group", "Residuals"),
  SumOfSquares = round(anovaTable$`Sum Sq`, 3),
  df = anovaTable$Df,
  MeanSquare = round(anovaTable$`Mean Sq`, 3),
  F = c(round(anovaTable$`F value`[1], 3), ""),
  p = c(round(anovaTable$`Pr(>F)`[1], 4), "")
)
print("ANOVA Table")
write.table(anovaPrint, sep = " | ", quote = FALSE, row.names = FALSE)
cat("\n")

#Post Hoc Comparisons
tukeyResults <- TukeyHSD(anovaModel)
tukeyData <- as.data.frame(tukeyResults$group)

tukeyPrint <- data.frame(
  Comparison = rownames(tukeyData),
  MeanDifference = round(tukeyData$diff, 3),
  LowerCI = round(tukeyData$lwr, 3),
  UpperCI = round(tukeyData$upr, 3),
  padj = round(tukeyData$`p adj`, 3)
)
print("Post Hoc Comparisons")
write.table(tukeyPrint, sep = " | ", quote = FALSE, row.names = FALSE)
```
