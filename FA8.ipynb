```r
Participant <- 1:24
Cloak <- c(rep(0, 12), rep(1, 12))
Mischief <- c(3, 1, 5, 4, 6, 4, 6, 2, 0, 5, 4, 5, 4, 3, 6, 6, 8, 5, 5, 4, 2, 5, 7, 5)

invisibility_data <- data.frame(Participant, Cloak, Mischief)

invisibility_data[["Group"]] <- factor(invisibility_data[["Cloak"]],
                                  levels = c(0, 1),
                                  labels = c("Without a cloak", "With a cloak"))

cat("INVISIBILITY CLOAK: INDEPENDENT SAMPLES T-TEST\n\n")

cat("ASSUMPTION 1: Continuous Dependent Variable\n")
cat("The dependent variable, Mischief, is measured at a continuous level because it is a direct count of the total number of mischievous acts committed. This numeric score acts as a continuous scale.\n\n")

cat("ASSUMPTION 2: Two Categorical Independent Groups\n")
cat("The independent variable, Cloak, consists of exactly two categorical, independent groups. The participants are cleanly divided into either having a cloak or not having a cloak.\n\n")

cat("ASSUMPTION 3: Independence of Observations\n")
cat("The observations are independent because each participant is present in only one group. Nobody was tested under both the cloak and no-cloak conditions.\n\n")

cat("ASSUMPTION 4: Outliers\n")
cat("Based on the visual inspection of the generated boxplot, there are no significant outliers. There are no extreme data points sitting far outside the main boxes.\n\n")

boxplot(invisibility_data[["Mischief"]] ~ invisibility_data[["Group"]],
        main = "Boxplot of Mischief by Cloak Group",
        ylab = "Number of Mischievous Acts",
        xlab = "Cloak Condition",
        col = c("#E69F00", "#56B4E9"))

cat("ASSUMPTION 5: Normality\n")
cat("The dependent variable needs to be normally distributed for each group, which is assessed by the Shapiro-Wilk test. A p-value greater than 0.05 indicates the data is normal.\n")

shapiro_without <- shapiro.test(invisibility_data[["Mischief"]][invisibility_data[["Group"]] == "Without a cloak"])
shapiro_with <- shapiro.test(invisibility_data[["Mischief"]][invisibility_data[["Group"]] == "With a cloak"])

cat("- Without a cloak p-value:", round(shapiro_without[["p.value"]], 3), "\n")
cat("- With a cloak p-value:", round(shapiro_with[["p.value"]], 3), "\n")
cat("Since both p-values are above 0.05, the data does not deviate from normality.\n\n")


cat("COMPUTATION: Independent Samples T-Test\n")
t_test_result <- t.test(invisibility_data[["Mischief"]] ~ invisibility_data[["Group"]], var.equal = TRUE)
print(t_test_result)
cat("\n")

cat("SHORT APA REPORT\n")
cat("An independent-samples t-test was conducted to examine the adequacy of the null hypothesis that the Invisibility Cloak has no effect on the number of mischievous acts committed. ")
cat("The differences between the groups are not statistically significant at the .05 level (t(22) = -1.713, p = 0.101). ")
cat("This means that the data do not provide much motivation to reject the null hypothesis. ")
cat("Therefore, there is no significant difference between With a Cloak and Without a Cloak on their number of mischievous acts.\n")
```
