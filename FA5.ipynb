```r
#NO.1
set.seed(123)

deck <- c(rep("King", 4), rep("NonKing", 48))
n <- 100000
draws <- replicate(n, sample(deck, 3, replace = FALSE))

E1 <- draws[1, ] == "King"
E2 <- draws[2, ] == "King"
E3 <- draws[3, ] == "King"

cat("NUMBER 1\n\n")

cat("(a) Probability that the first and second draws are not kings:\n")
cat("    =", mean(!E1 & !E2), "\n\n")

cat("(b) Probability that a king appears on the first or second draw:\n")
cat("    =", mean(E1 | E2), "\n\n")

cat("(c) Probability that neither the first nor second draw is a king:\n")
cat("    =", mean(!(E1 | E2)), "\n\n")

cat("(d) Probability that the third draw is a king given the first two are not kings:\n")
cat("    =", mean(E3[!E1 & !E2]), "\n\n")

cat("(e) Probability that first is not king, second and third are kings:\n")
cat("    =", mean(!E1 & E2 & E3), "\n\n")

cat("(f) Probability of two consecutive kings:\n")
cat("    =", mean((E1 & E2) | (E2 & E3)), "\n\n")

#NO.2
outcomes <- expand.grid(d1 = 1:6, d2 = 1:6, d3 = 1:6)
X <- outcomes$d1 + outcomes$d2 + outcomes$d3

prob_dist <- table(X) / length(X)

values <- as.numeric(names(prob_dist))
probabilities <- as.numeric(prob_dist)

dist_table <- data.frame(
  X = values,
  P_of_X = round(probabilities, 6)
)

cat("NUMBER 2\n")
cat("(a)\n")
print(dist_table)
cat("\n")

cat("(b) =", round(sum(probabilities[values >= 7 & values <= 11]), 6), "\n\n")


#NO.3
x <- c(-10, -20, 30)
p <- c(1/5, 3/10, 1/2)

EX <- sum(x * p)
EX2 <- sum(x^2 * p)
VarX <- sum((x - EX)^2 * p)
EX3 <- sum(x^3 * p)

cat("NUMBER 3\n")
cat("(a) =", EX, "\n")
cat("(b) =", EX2, "\n")
cat("(c) =", VarX, "\n")
cat("(d) =", EX3, "\n")


#NO.4
cat("\nNUMBER 4\n")
cat("(a) =", factorial(6 - 1), "\n")

total <- factorial(5)
block <- factorial(4) * 2
cat("(b) =", total - block, "\n")

men <- factorial(3 - 1)
women <- factorial(3)
cat("(c) =", men * women, "\n\n")


#NO.5
cat("NUMBER 5\n")
cat("(a) =", choose(6, 2) * choose(8, 4) * choose(4, 3) * choose(5, 3), "\n")

cat("(b) =", choose(5, 1) * choose(7, 3) * choose(4, 3) * choose(5, 3), "\n\n")


#NO.6
cat("NUMBER 6\n")

cat("Addition Rule:\n")
cat("P(E1 ∪ E2) = P(E1) + P(E2) - P(E1 ∩ E2)\n\n")

cat("Since probabilities are non-negative:\n")
cat("P(E1 ∩ E2) >= 0\n\n")

cat("Therefore:\n")
cat("P(E1) + P(E2) - P(E1 ∩ E2) <= P(E1) + P(E2)\n\n")

cat("Hence:\n")
cat("P(E1 ∪ E2) <= P(E1) + P(E2)\n")
```
