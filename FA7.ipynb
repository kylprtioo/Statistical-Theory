```r
#9.21
masses <- c(8.3, 10.6, 9.7, 8.8, 10.2, 9.4)
n <- length(masses)

estpopmean <- mean(masses)

estpopvar <- var(masses)

estpopsd <- sd(masses)
samplesd <- sqrt(sum((masses - estpopmean)^2) / n)

print("9.21")
cat("(a) Unbiased estimate of population mean:", estpopmean, "\n")
cat("(b) Unbiased estimate of population variance:", estpopvar, "\n")
cat("(c) Estimated Population SD:", estpopsd, "\n")
cat("    Sample SD:", samplesd, "\n")
```

```r
#9.24
n <- 60
xbar <- 11.09
s <- 0.73

se <- s / sqrt(n)

z95 <- qnorm(0.975)
margin95 <- z95 * se
lower95 <- xbar - margin95
upper95 <- xbar + margin95

z99 <- qnorm(0.995)
margin99 <- z99 * se
lower99 <- xbar - margin99
upper99 <- xbar + margin99

print("9.24")
cat("(a) 95% Confidence Limits:", round(lower95, 3), "to", round(upper95, 3), "\n")
cat("(b) 99% Confidence Limits:", round(lower99, 3), "to", round(upper99, 3), "\n")
```

```r
#9.33
n1 <- 50
xbar1 <- 7.82
s1 <- 0.24

n2 <- 100
xbar2 <- 6.75
s2 <- 0.30

diffmean <- xbar1 - xbar2
sediff <- sqrt((s1^2 / n1) + (s2^2 / n2))

z95 <- qnorm(0.975)
me95 <- z95 * sediff
lower95 <- diffmean - me95
upper95 <- diffmean + me95

z99 <- qnorm(0.995)
me99 <- z99 * sediff
lower99 <- diffmean - me99
upper99 <- diffmean + me99

print("9.33")
cat("(a) 95% Confidence Limits:", round(lower95, 3), "to", round(upper95, 3), "\n")
cat("(b) 99% Confidence Limits:", round(lower99, 3), "to", round(upper99, 3), "\n")
```

```r
#9.34
males <- c(85, 60, 55, 90, 49, 90, 62, 78, 53, 72, 53, 51, 99, 72, 72, 74, 65, 49, 82, 104, 100, 61, 56, 62, 58, 85, 81, 78, 109, 70, 49, 83, 55, 69, 60, 80, 55, 80, 87, 31, 65, 65, 55, 59, 68, 77, 71, 75, 78, 50)

females <- c(64, 64, 61, 105, 71, 98, 66, 81, 92, 91, 93, 60, 84, 90, 99, 61, 74, 86, 77, 93, 82, 75, 91, 59, 98, 108, 60, 65, 82, 63, 71, 87, 61, 86, 54, 79, 90, 86, 86, 93, 77, 60, 85, 62, 94, 50, 95, 81, 79, 53)

nm <- length(males)
meanm <- mean(males)
sdm <- sd(males)

nf <- length(females)
meanf <- mean(females)
sdf <- sd(females)

diffmean <- meanm - meanf
se <- sqrt((sdm^2 / nm) + (sdf^2 / nf))

z85 <- qnorm(0.925)
me85 <- z85 * se

lower85 <- diffmean - me85
upper85 <- diffmean + me85

print("9.34")
cat("Mean Males:", round(meanm, 2), "| SD Males:", round(sdm, 2), "\n")
cat("Mean Females:", round(meanf, 2), "| SD Females:", round(sdf, 2), "\n")
cat("Point Estimate of Difference:", round(diffmean, 2), "\n")
cat("85% Confidence Interval:", round(lower85, 3), "to", round(upper85, 3), "\n")
```
